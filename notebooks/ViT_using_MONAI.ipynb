## Imports

```python
import logging
import numpy as np
import os
from pathlib import Path
import sys
import tempfile
import torch
import ignite

from monai.config import print_config
from monai.data import DataLoader
from monai.engines import SupervisedTrainer, SupervisedEvaluator
from monai.handlers import (
    StatsHandler,
    TensorBoardStatsHandler,
    ValidationHandler,
    CheckpointSaver,
    CheckpointLoader,
    ClassificationSaver,
)
from monai.handlers.utils import from_engine
from monai.inferers import SimpleInferer
from monai.networks.nets import ViT
from monai.transforms import LoadImageD, EnsureChannelFirstD, ScaleIntensityD, Compose, AsDiscreted
from monai.data import Dataset
from monai.transforms import (
    Compose,
    LoadImaged,
    EnsureChannelFirstd,
    Resized,
    ScaleIntensityd,
    EnsureTyped,
)

print_config()
```

## Loading Dataset

```python
DATA_DIR = Path(
    r"..\data\brain-tumor-mri-classification\merged_dataset_v3\images"
)
```

```python
class_names = [
    "glioma",
    "meningioma",
    "no_tumor",
    "pituitary",
]

class_to_idx = {
    class_name: idx
    for idx, class_name in enumerate(class_names)
}
```

```python
def get_image_files(split):
    split_dir = DATA_DIR / split
    data = []

    for class_name in class_names:
        class_dir = split_dir / class_name

        for image_path in class_dir.iterdir():
            if image_path.suffix.lower() in [".jpg", ".jpeg", ".png"]:
                data.append(
                    {
                        "image": str(image_path),
                        "label": class_to_idx[class_name],
                    }
                )

    return data
```

```python
transforms = Compose([
    LoadImaged(keys=["image"]),
    EnsureChannelFirstd(keys=["image"]),
    Resized(keys=["image"], spatial_size=(224, 224)),
    ScaleIntensityd(keys=["image"]),
    EnsureTyped(keys=["image"]),
])
```

```python
train_files = get_image_files("train")
val_files = get_image_files("val")
test_files = get_image_files("test")
```

```python
train_dataset = Dataset(
    data=train_files,
    transform=transforms,
)

val_dataset = Dataset(
    data=val_files,
    transform=transforms,
)

test_dataset = Dataset(
    data=test_files,
    transform=transforms,
)
```

```python
print("Train images:", len(train_dataset))
print("Validation images:", len(val_dataset))
print("Test images:", len(test_dataset))

print("First sample:", train_dataset[0])
```

## Select subset of dataset

```python
from torch.utils.data import Subset
import random

random.seed(42)

subset_size = 8306

train_indices = random.sample(
    range(len(train_dataset)),
    min(subset_size, len(train_dataset))
)

val_indices = random.sample(
    range(len(val_dataset)),
    min(subset_size, len(val_dataset))
)

test_indices = random.sample(
    range(len(test_dataset)),
    min(subset_size, len(test_dataset))
)

train_subset = Subset(train_dataset, train_indices)
val_subset = Subset(val_dataset, val_indices)
test_subset = Subset(test_dataset, test_indices)

print("Training images:", len(train_subset))
print("Validation images:", len(val_subset))
print("Test images:", len(test_subset))
```

## Training Model

```python
max_epochs = 50
save_interval = 2
out_dir = "../eval/vit"

device = torch.device("cuda:0")

class ViTWrapper(torch.nn.Module):
    def __init__(self, vit_model):
        super().__init__()
        self.vit_model = vit_model

    def forward(self, x):
        outputs = self.vit_model(x)
        # Return only the first element (logits) if it's a tuple
        if isinstance(outputs, tuple):
            return outputs[0]
        return outputs

base_model = ViT(in_channels=3, img_size=(224, 224), patch_size=(16, 16), num_classes=4, spatial_dims=2, classification=True)
model = ViTWrapper(base_model).to(device)

loss_fn = torch.nn.CrossEntropyLoss()

logging.basicConfig(
    stream=sys.stdout,
    level=logging.INFO,
)

# --------------------------------------------------
# VALIDATION METRICS
# --------------------------------------------------

evaluator = SupervisedEvaluator(
    device=device,
    val_data_loader=DataLoader(
        val_dataset,
        batch_size=512,
        shuffle=False,
        num_workers=0,
    ),
    network=model,
    inferer=SimpleInferer(),
    key_val_metric={
        "val_accuracy": ignite.metrics.Accuracy(
            output_transform=from_engine(["pred", "label"])
        ),

        "val_loss": ignite.metrics.Loss(
            loss_fn=loss_fn,
            output_transform=from_engine(["pred", "label"]),
        ),
    },

    val_handlers=[
        StatsHandler(
            iteration_log=False,
        ),

        TensorBoardStatsHandler(
            iteration_log=False,
            tag_name="validation",
        ),
    ],
)

# --------------------------------------------------
# TRAINING
# --------------------------------------------------

trainer = SupervisedTrainer(
    device=device,
    max_epochs=max_epochs,

    train_data_loader=DataLoader(
        train_subset,
        batch_size=32,
        shuffle=True,
        num_workers=0,
    ),

    network=model,

    optimizer=torch.optim.Adam(
        model.parameters(),
        lr=1e-5,
    ),

    loss_function=loss_fn,
    inferer=SimpleInferer(),

    key_train_metric={
        "train_accuracy": ignite.metrics.Accuracy(
            output_transform=from_engine(["pred", "label"])
        ),
        "train_loss": ignite.metrics.Loss(
            loss_fn=loss_fn,
            output_transform=from_engine(["pred", "label"]),
        ),
    },

    train_handlers=[
        ValidationHandler(
            validator=evaluator,
            epoch_level=True,
            interval=1,
        ),

        CheckpointSaver(
            save_dir=out_dir,
            save_dict={"model": model},
            save_interval=save_interval,
            save_final=True,
            final_filename="monai_vit.pt",
        ),

        # StatsHandler(),

        TensorBoardStatsHandler(
            tag_name="train",
            output_transform=from_engine(
                ["loss"],
                first=True,
            ),
        ),
    ],
)
```

```python
from ignite.engine import Events
import datetime

train_losses = []
val_losses = []
train_accuracies = []
val_accuracies = []

@trainer.on(Events.EPOCH_COMPLETED)
def log_epoch_metrics(engine):
    epoch = engine.state.epoch

    train_accuracy = engine.state.metrics.get("train_accuracy")
    train_loss = engine.state.metrics.get("train_loss")

    val_accuracy = evaluator.state.metrics.get("val_accuracy")
    val_loss = evaluator.state.metrics.get("val_loss")

    current_time = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    epoch_duration = engine.state.times.get(Events.EPOCH_COMPLETED.name, 0.0)

    train_losses.append(train_loss)
    val_losses.append(val_loss)
    train_accuracies.append(train_accuracy)
    val_accuracies.append(val_accuracy)


    print(
        f"\n {current_time} | "
        f"Epoch {epoch}/{max_epochs} | "
        f"Train Accuracy: {train_accuracy} | "
        f"Train Loss: {train_loss} | "
        f"Val Accuracy: {val_accuracy} | "
        f"Val Loss: {val_loss} | "
        f"Time taken: {epoch_duration}\n"
    )
```

```python
trainer.run()
```

```python
import matplotlib.pyplot as plt

epochs = range(1, max_epochs + 1)

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(epochs, train_losses, marker="o", label="Train Loss")
plt.plot(epochs, val_losses, marker="o", label="Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Loss Curves")
plt.legend()
plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(epochs, train_accuracies, marker="o", label="Train Accuracy")
plt.plot(epochs, val_accuracies, marker="o", label="Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("Accuracy Curves")
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
import torch
from monai.data import DataLoader

test_loader = DataLoader(
    test_subset,
    batch_size=32,
    shuffle=False,
    num_workers=0,
)

model.eval()

correct = 0
total = 0

with torch.no_grad():
    for batch in test_loader:
        images = batch["image"].to(device)
        labels = batch["label"].to(device)

        logits = model(images)

        predictions = logits.argmax(dim=1)

        correct += (predictions == labels).sum().item()
        total += labels.size(0)

test_accuracy = correct / total

print(f"Test Accuracy: {test_accuracy * 100:.2f}%")
```

## Evaluation

```python
import torch
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from torch.utils.data import DataLoader
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
)
```

```python
test_loader = DataLoader(
    test_dataset,
    batch_size=32,
    shuffle=False,
    num_workers=0,
)
```

```python
device = torch.device("cuda:0" if torch.cuda.is_available() else "cpu")
model = model.to(device)
model.eval()
```

```python
all_labels = []
all_predictions = []

with torch.inference_mode():
    for batch in test_loader:
        images = batch["image"].to(device)
        labels = batch["label"].to(device)

        # Model output: [batch_size, 4]
        logits = model(images)

        # Convert logits into predicted class indices
        predictions = torch.argmax(logits, dim=1)

        all_labels.extend(labels.cpu().numpy())
        all_predictions.extend(predictions.cpu().numpy())

all_labels = np.array(all_labels)
all_predictions = np.array(all_predictions)
```

### Accuracy, Precision, Recall, F1 Score

```python
accuracy = accuracy_score(all_labels, all_predictions)

precision = precision_score(
    all_labels,
    all_predictions,
    average="weighted",
    zero_division=0,
)

recall = recall_score(
    all_labels,
    all_predictions,
    average="weighted",
    zero_division=0,
)

f1 = f1_score(
    all_labels,
    all_predictions,
    average="weighted",
    zero_division=0,
)

print(f"Test Accuracy : {accuracy:.4f}")
print(f"Test Precision: {precision:.4f}")
print(f"Test Recall   : {recall:.4f}")
print(f"Test F1 Score : {f1:.4f}")
```

### Classificaiton Report

```python
print("\nClassification Report:\n")

print(
    classification_report(
        all_labels,
        all_predictions,
        labels=list(range(len(class_names))),
        target_names=class_names,
        zero_division=0,
    )
)
```

### Confusion Matrix

```python
cm = confusion_matrix(
    all_labels,
    all_predictions,
    labels=list(range(len(class_names))),
)

plt.figure(figsize=(8, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    xticklabels=class_names,
    yticklabels=class_names,
    cmap="Blues"
)

plt.xlabel("Predicted Label")
plt.ylabel("True Label")
plt.title("Confusion Matrix - Brain Tumor Classification")
plt.tight_layout()
plt.show()
```
