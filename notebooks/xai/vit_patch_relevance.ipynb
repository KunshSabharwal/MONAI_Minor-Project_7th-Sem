## Imports

```python
import os
import random
import numpy as np
import matplotlib.pyplot as plt

import torch
import torch.nn.functional as F
import timm

from monai.transforms import (
    Compose,
    LoadImage,
    EnsureChannelFirst,
    Resize,
    ScaleIntensity,
    EnsureType,
)
```

## Config + Check GPU

```python
PROJECT_DIR = r"C:\Users\user\Desktop\MONAI"

MODEL_PATH = os.path.join(
    PROJECT_DIR,
    "models",
    "vit_base_patch16_224.pt"
)

DATA_DIR = os.path.join(
    PROJECT_DIR,
    "data",
    "brain-tumor-mri-classification",
    "merged_dataset_v3",
    "images"
)

TEST_DIR = os.path.join(DATA_DIR, "test")

class_names = [
    "glioma",
    "meningioma",
    "no_tumor",
    "pituitary"
]

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Device:", device)

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
```

## Load Model

```python
model = timm.create_model(
    "vit_base_patch16_224",
    pretrained=False,
    num_classes=4
)

model = model.to(device)
```

```python
checkpoint = torch.load(
    MODEL_PATH,
    map_location=device
)

print(type(checkpoint))

if isinstance(checkpoint, dict):
    print(checkpoint.keys())
```

```python
model.load_state_dict(checkpoint)

model.eval()

print("ViT loaded successfully!")
```

```python
transform = Compose([
    LoadImage(image_only=True),
    EnsureChannelFirst(),
    Resize((224, 224)),
    ScaleIntensity(),
    EnsureType(),
])
```

## Select 8 random images

```python
random.seed(42)

selected_images = []

for class_name in class_names:

    class_dir = os.path.join(
        TEST_DIR,
        class_name
    )

    images = [
        os.path.join(class_dir, f)
        for f in os.listdir(class_dir)
        if f.lower().endswith(
            (".jpg", ".jpeg", ".png", ".bmp")
        )
    ]

    selected = random.sample(
        images,
        min(2, len(images))
    )

    for image_path in selected:
        selected_images.append({
            "path": image_path,
            "true_class": class_name
        })

print("Selected:", len(selected_images))

for item in selected_images:
    print(
        item["true_class"],
        "->",
        os.path.basename(item["path"])
    )
```

```python
target_layer = model.blocks[-1].norm1

print(target_layer)
```

## Apply GRAD CAM

```python
def generate_vit_xai(model, image_path, target_layer):

    activations = None
    gradients = None

    def forward_hook(module, inputs, output):
        nonlocal activations
        activations = output

    def backward_hook(module, grad_input, grad_output):
        nonlocal gradients
        gradients = grad_output[0]

    forward_handle = target_layer.register_forward_hook(
        forward_hook
    )

    backward_handle = target_layer.register_full_backward_hook(
        backward_hook
    )

    # -----------------------------------------
    # Load image
    # -----------------------------------------

    image = transform(image_path)

    input_tensor = image.unsqueeze(0).to(device)

    # -----------------------------------------
    # Forward
    # -----------------------------------------

    model.zero_grad()

    output = model(input_tensor)

    probabilities = F.softmax(output, dim=1)

    predicted_class = output.argmax(dim=1).item()

    confidence = probabilities[
        0, predicted_class
    ].item()

    # -----------------------------------------
    # Backward
    # -----------------------------------------

    score = output[
        0, predicted_class
    ]

    score.backward()

    # -----------------------------------------
    # Get activations + gradients
    # -----------------------------------------

    activations = activations.detach()
    gradients = gradients.detach()

    print("Activations:", activations.shape)
    print("Gradients:", gradients.shape)

    # -----------------------------------------
    # Remove CLS token
    # -----------------------------------------

    activations = activations[:, 1:, :]
    gradients = gradients[:, 1:, :]

    # Shape:
    # [1, 196, 768]

    # -----------------------------------------
    # Patch importance
    #
    # Each patch gets one importance value.
    # -----------------------------------------

    cam = (
        gradients * activations
    ).sum(dim=-1)

    # Shape:
    # [1, 196]

    cam = cam.squeeze(0)

    # -----------------------------------------
    # Keep positive contributions
    # -----------------------------------------

    cam = F.relu(cam)

    # -----------------------------------------
    # 196 patches -> 14 x 14
    # -----------------------------------------

    cam = cam.reshape(14, 14)

    # -----------------------------------------
    # Normalize
    # -----------------------------------------

    cam_min = cam.min()
    cam_max = cam.max()

    if cam_max > cam_min:
        cam = (
            cam - cam_min
        ) / (
            cam_max - cam_min
        )
    else:
        cam = torch.zeros_like(cam)

    # -----------------------------------------
    # Resize patch map -> 224 x 224
    # -----------------------------------------

    cam = F.interpolate(
        cam.unsqueeze(0).unsqueeze(0),
        size=(224, 224),
        mode="bilinear",
        align_corners=False
    )

    cam = cam.squeeze().cpu().numpy()

    # -----------------------------------------
    # Prepare image
    # -----------------------------------------

    display_image = (
        image
        .detach()
        .cpu()
        .numpy()
        .transpose(1, 2, 0)
    )

    display_image = (
        display_image - display_image.min()
    )

    if display_image.max() > 0:
        display_image = (
            display_image /
            display_image.max()
        )

    # -----------------------------------------
    # Remove hooks
    # -----------------------------------------

    forward_handle.remove()
    backward_handle.remove()

    return (
        display_image,
        cam,
        predicted_class,
        confidence
    )
```

```python
results = []

for item in selected_images:

    print(
        f"Processing: {item['true_class']} -> "
        f"{os.path.basename(item['path'])}"
    )

    (
        image,
        cam,
        predicted_class,
        confidence
    ) = generate_vit_xai(
        model,
        item["path"],
        target_layer
    )

    results.append({
        "path": item["path"],
        "true_class": item["true_class"],
        "predicted_class": class_names[predicted_class],
        "confidence": confidence,
        "image": image,
        "cam": cam
    })

print("\nCompleted:", len(results), "images")
```

```python
fig, axes = plt.subplots(
    8,
    3,
    figsize=(13, 30)
)

for i, result in enumerate(results):

    image = result["image"]
    cam = result["cam"]

    true_class = result["true_class"]
    predicted_class = result["predicted_class"]
    confidence = result["confidence"]

    # --------------------------------
    # Original
    # --------------------------------

    axes[i, 0].imshow(image)

    axes[i, 0].set_title(
        f"Original\n"
        f"True: {true_class}"
    )

    axes[i, 0].axis("off")

    # --------------------------------
    # XAI heatmap
    # --------------------------------

    axes[i, 1].imshow(
        cam,
        cmap="jet"
    )

    axes[i, 1].set_title(
        "ViT Patch Relevance"
    )

    axes[i, 1].axis("off")

    # --------------------------------
    # Overlay
    # --------------------------------

    axes[i, 2].imshow(image)

    axes[i, 2].imshow(
        cam,
        cmap="jet",
        alpha=0.5
    )

    axes[i, 2].set_title(
        f"Prediction: {predicted_class}\n"
        f"Confidence: {confidence:.2%}"
    )

    axes[i, 2].axis("off")


plt.tight_layout()
plt.show()
```
